## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This task is fundamentally a ranking and scoring problem. While we may utilize classification algorithms (like a decision tree or random forest) to calculate the probability of a page declining, the final required output is a priority score used to rank candidates. The decision we are improving is "Which ones first?", specifically, ordering the queue so human reviewers address the highest-value opportunities before their capacity is exhausted.

```python
task_type = "Ranking / Scoring"
primary_output = "Ranked priority queue of content for human review"
print(f"ML Task Frame: {task_type}")
print(f"Deliverable: {primary_output}")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is a proxy: `is_declining_label`, which is derived from the `trend_direction` rule. However, training on a defined rule means the model only learns the rule, not the real world. To build a production-grade system, the ultimate target must be an observed outcome. We must predict a measurable, future reality; such as a quantifiable drop in traffic or sessions occurring over a subsequent time window

```python
starter_proxy_target = "is_declining_label (Defined Rule)"
ideal_observed_target = "Subsequent traffic drop (Observed Outcome)"

print(f"Starter Target (Proxy): {starter_proxy_target}")
print(f"Ideal Capstone Target: {ideal_observed_target}")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric is precision@K, where K represents the human review team's actual capacity (for example: precision@50). A "good" number means that out of the top 50 pages the system recommends, a high percentage actually exhibited the target behavior (a true drop in performance or a valid refresh opportunity). This strictly measures our ability to minimize the cost of a wrong call, which is wasted editor hours

```python
capacity_k = 50
success_metric = f"Precision@{capacity_k}"
definition_of_good = f"High percentage of true positives within the top {capacity_k} ranked pages"

print(f"Optimization Metric: {success_metric}")
print(f"Operational Goal: {definition_of_good}")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (if-statements) is insufficient because the data patterns are tangled and shift depending on context. For instance, a Click-Through Rate (CTR) of 1% might be excellent for a page at position 9, but catastrophic for a page at position 1.

Machine learning algorithms can balance these interacting, non-linear variables—such as age, search volume, position, and engagement rates—simultaneously. If we tried to write this by hand, the nested if-statements would become brittle, unmaintainable, and easily broken by edge cases.

```python
# Demonstrating the variance that breaks static if-statements
if 'ctr' in df.columns and 'avg_position' in df.columns:
    # Remember: avg_position = 0 means 'no data', not rank zero
    valid_positions = df[df['avg_position'] > 0]
    print("Variance in CTR across different positions makes static rules ineffective:")
    print(valid_positions[['avg_position', 'ctr']].describe().loc[['mean', 'std', 'min', 'max']])
```
